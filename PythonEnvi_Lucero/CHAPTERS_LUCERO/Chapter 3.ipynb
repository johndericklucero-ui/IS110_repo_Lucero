#### Exercise 1

In the code below, which will never print regardless of the value for x? Edit the code to correct the error.
# they are not in order
```python
if x > 0 :
    print('Above 0')
elif x >  10 :
    print('Above 10')
elif x > 50 :
    print('Above 50')
elif x >  20 :
    print('Above 20')
else :
    print('Something else')
```

````python

def check_x_value(x):
    if x > 50:
        print('Above 50')
    elif x > 20:
        print('Above 20')
    elif x > 10:
        print('Above 10')
    elif x > 0:
        print('Above 0')
    else:
        print('Something else')

print("Testing with x = 60:")
check_x_value(60)

print("\nTesting with x = 25:")
check_x_value(25)

print("\nTesting with x = 15:")
check_x_value(15)

print("\nTesting with x = 5:")
check_x_value(5)

print("\nTesting with x = -5:")
check_x_value(-5)
````

#### Exercise 2

Write your salary computation to give an employee 1.5 times the hourly rate for hours worked above 40 hours. Example:

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    
[PEW] E3.2

````python
try:
    hours = float(input('Enter Hours: '))
    rate = float(input('Enter Rate: '))

    if hours > 40:
        pay = (40 * rate) + ((hours - 40) * rate * 1.5)
    else:
        pay = hours * rate

    print('Pay:', pay)

except ValueError:
    print("Error: Please enter numeric input for hours and rate.")
````

## CHALLENGE

Write a program to prompt for a score between 0 and 100 and print a grade using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		3.0

Adapted from [PEW] E3.3
<br>

````python
try:
    score_str = input('Enter a score between 0 and 100: ')
    score = float(score_str)

    if not (0 <= score <= 100):
        print('Error: Score must be between 0 and 100.')
    elif score >= 98:
        print('Grade: 1.0')
    elif score >= 94:
        print('Grade: 1.25')
    elif score >= 90:
        print('Grade: 1.5')
    elif score >= 86:
        print('Grade: 1.75')
    elif score >= 82:
        print('Grade: 2.0')
    elif score >= 78:
        print('Grade: 2.25')
    elif score >= 75:
        print('Grade: 2.5')
    else:
        print('Grade: 3.0')
except ValueError:
    print('Error: Please enter a numeric score.')
````
