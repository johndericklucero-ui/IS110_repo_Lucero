### Part A: Creating an `ndarray`

```python
#values I used 1 - 9
python_list_A = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

A = np.array(python_list_A)

print("Our NumPy array A:")
print(A)
```

#### 1. Using a property of the `ndarray` object (`A.shape`)

```python
# The 'shape' property returns a tuple indicating the size of each dimension of the array.
# For a 3x3 array, it will return (3, 3).
print("Shape of array A:", A.shape)
```

#### 2. Using a method of the `ndarray` object (`A.max()`)

```python
# The 'max()' method returns the maximum value present in the array inside of it.
print("Maximum value in array A :", A.max())
```

#### 3. Using a NumPy function and passing an `ndarray` (`np.max(A)`)

```python
print("Maximum value in array A (using NumPy function):", np.max(A))
```

### Part B: Create a 5x5 integer-array filled with ones and check its properties

```python
#valus will increment

B_array = np.ones((5, 5), dtype=int)
print("The 5x5 integer array filled with ones:")
print(B_array)

# This will return a tuple indicating the dimensions (rows, columns).

print("\nShape of B_array:", B_array.shape)

# This will return the total number of elements in the array (rows * columns).

print("Size of B_array:", B_array.size)

# This will return an integer representing the number of dimensions.

print("Number of dimensions of B_array:", B_array.ndim)
```

### Part C: Create an array from -5 to +5, multiply by 2, and print

```python
# Create an array of values from -5 to 5 using np.arange()
# np.arange(-5, 6) generates numbers from -5 up to (but not including) 6, which covers -5 to 5.

C_array = np.arange(-5, 6)

print("Original array (C_array):")
print(C_array)

# Multiply each value in the array by 2
# NumPy arrays support element-wise operations, so C_array * 2 multiplies each element by 2.

C_array_multiplied = C_array * 2

print("\nArray after multiplying each value by 2:")
print(C_array_multiplied)
```
